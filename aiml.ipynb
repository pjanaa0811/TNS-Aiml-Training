```python
import numpy as np

# Marks of 10 students
marks = np.array([85, 72, 91, 68, 77, 95, 64, 81, 73, 88])

print("Marks:", marks)

# Total marks
total = np.sum(marks)
print("Total Marks:", total)

# Average marks
average = np.mean(marks)
print("Average Marks:", average)

# Highest and lowest marks
highest = np.max(marks)
lowest = np.min(marks)

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)

# Marks greater than 75
greater_than_75 = marks[marks > 75]
print("Marks greater than 75:", greater_than_75)
```

```python
import numpy as np

# Temperature for 7 days
temperatures = np.array([28, 31, 29, 33, 35, 30, 32])

print("Temperatures:", temperatures)

# Average temperature
average = np.mean(temperatures)
print("Average Temperature:", average)

# Highest and lowest temperature
highest = np.max(temperatures)
lowest = np.min(temperatures)

print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)

# Days where temperature is above 30°C
days = np.where(temperatures > 30)[0] + 1

print("Days with temperature above 30°C:", days)

# Increase all temperatures by 2°C
updated_temperatures = temperatures + 2

print("Updated Temperatures:", updated_temperatures)
```

```python
import pandas as pd

# Student data
data = {
    "Name": ["Arun", "Bala", "Charan", "Divya", "Ezhil",
             "Farhan", "Gokul", "Hari"],

    "Department": ["IT", "CSE", "ECE", "IT", "CSE",
                   "ECE", "IT", "CSE"],

    "Marks": [85, 72, 91, 68, 78, 95, 64, 82],

    "Attendance": [90, 85, 92, 75, 88, 95, 70, 82]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Student Data:")
print(df)

# Display first 5 students
print("\nFirst 5 Students:")
print(df.head())

# Average marks
average_marks = df["Marks"].mean()
print("\nAverage Marks:", average_marks)

# Students who scored more than 75
print("\nStudents who scored more than 75:")
print(df[df["Marks"] > 75])

# Students whose attendance is below 80%
print("\nStudents with attendance below 80%:")
print(df[df["Attendance"] < 80])

# Sort students based on marks
print("\nStudents sorted by marks:")
print(df.sort_values(by="Marks"))
```

```python
import pandas as pd

# Product data
data = {
    "Product Name": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Category": ["Electronics", "Electronics", "Accessories",
                 "Accessories", "Accessories"],
    "Price": [50000, 25000, 2000, 1500, 800],
    "Quantity Sold": [20, 40, 75, 60, 100]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Product Data:")
print(df)

# Calculate total sales amount
df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("\nTotal Sales for Each Product:")
print(df[["Product Name", "Total Sales"]])

# Product with highest sales
highest_sales_product = df.loc[df["Total Sales"].idxmax()]

print("\nProduct with Highest Sales:")
print(highest_sales_product)

# Average product price
average_price = df["Price"].mean()

print("\nAverage Product Price:", average_price)

# Products where quantity sold is greater than 50
print("\nProducts with Quantity Sold greater than 50:")
print(df[df["Quantity Sold"] > 50])

# Sort products based on total sales
print("\nProducts Sorted by Total Sales:")
print(df.sort_values(by="Total Sales", ascending=False))
```

```python
import matplotlib.pyplot as plt

# Monthly sales data
months = ["January", "February", "March", "April", "May", "June"]
sales = [25000, 30000, 28000, 35000, 40000, 45000]

# Line chart
plt.figure(figsize=(8, 5))

plt.plot(months, sales, marker="o", label="Monthly Sales")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales Amount")
plt.legend()
plt.grid(True)

plt.show()


# Bar chart
plt.figure(figsize=(8, 5))

plt.bar(months, sales, label="Monthly Sales")

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales Amount")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Student names and marks
students = ["Arun", "Bala", "Charan", "Divya", "Ezhil",
            "Farhan", "Gokul", "Hari", "Ishan", "Jaya"]

marks = [85, 72, 91, 68, 55, 95, 35, 82, 45, 76]

# -----------------------------
# Bar Chart
# -----------------------------

plt.figure(figsize=(10, 5))

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.xticks(rotation=45)

plt.show()


# -----------------------------
# Performance Categories
# -----------------------------

excellent = 0
good = 0
average = 0
needs_improvement = 0

for mark in marks:

    if 80 <= mark <= 100:
        excellent += 1

    elif 60 <= mark <= 79:
        good += 1

    elif 40 <= mark <= 59:
        average += 1

    else:
        needs_improvement += 1


# Category data
categories = [
    "Excellent",
    "Good",
    "Average",
    "Needs Improvement"
]

counts = [
    excellent,
    good,
    average,
    needs_improvement
]


# -----------------------------
# Pie Chart
# -----------------------------

plt.figure(figsize=(7, 7))

plt.pie(
    counts,
    labels=categories,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Distribution")

plt.show()
```
